```sql
-- q06: how often room rent goes over the plan's cap (per plan at admission)
SELECT
  p.plan_code,
  COUNT(*)                                        AS claim_count,
  SUM(CASE WHEN f.room_rent_over_cap THEN 1 ELSE 0 END) AS over_cap_count,
  ROUND(SUM(CASE WHEN f.room_rent_over_cap THEN 1 ELSE 0 END) / COUNT(*), 4) AS over_cap_share
FROM claims.analytics.fact_claims f
JOIN claims.analytics.dim_plan p ON f.plan_sk = p.plan_sk
GROUP BY p.plan_code
ORDER BY p.plan_code;
```
