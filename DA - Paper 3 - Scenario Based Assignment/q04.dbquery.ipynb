```sql
-- q04: how long settled claims take by claim type and network status at admission
SELECT
  f.claim_type,
  f.is_network_at_admission,
  COUNT(*)                        AS settled_claims,
  ROUND(AVG(f.turnaround_days),2) AS avg_turnaround_days
FROM claims.analytics.fact_claims f
WHERE f.status = 'settled'
GROUP BY f.claim_type, f.is_network_at_admission
ORDER BY f.claim_type, f.is_network_at_admission;
```
